```python
full_text = ""

for page in document:
    page_text = page.get_text()
    full_text += page_text + "\n"

print("Total characters:", len(full_text))
print("Preview:", full_text[:500])
```

```python
chunk_size = 1000
chunk_overlap = 200

chunks = []

for i in range(0, len(full_text), chunk_size - chunk_overlap):
    chunk = full_text[i:i + chunk_size]
    chunks.append(chunk)

print("Total chunks:", len(chunks))
print("First chunk:")
print(chunks[0][:300])
```

```python
from sentence_transformers import SentenceTransformer

# Load the embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Convert all chunks into embeddings
embeddings = model.encode(
    chunks,
    show_progress_bar=True
)

print("Embeddings shape:", embeddings.shape)
```

```python
import chromadb

# Create a persistent database
client = chromadb.PersistentClient(path="./chroma_db")

# Create or open our collection
collection = client.get_or_create_collection(
    name="documind_documents"
)

# Store chunks and their embeddings
collection.upsert(
    ids=[f"chunk_{i}" for i in range(len(chunks))],
    documents=chunks,
    embeddings=embeddings.tolist()
)

print("Stored chunks:", collection.count())
```

```python
# Our question
question = "How can family trauma affect future generations?"

# Convert the question into an embedding
question_embedding = model.encode(question)

# Search the vector database
results = collection.query(
    query_embeddings=[question_embedding.tolist()],
    n_results=3
)

# Display the 3 most relevant chunks
for i, text in enumerate(results["documents"][0]):
    print(f"\n--- Result {i + 1} ---")
    print(text[:500])
```

```python
# Day 2 — Step 1: Page-aware chunking

page_chunks = []
page_metadata = []

chunk_size = 1000
chunk_overlap = 200

for page_number, page in enumerate(document, start=1):
    page_text = page.get_text()

    if not page_text.strip():
        continue

    for i in range(0, len(page_text), chunk_size - chunk_overlap):
        chunk = page_text[i:i + chunk_size]

        if chunk.strip():
            page_chunks.append(chunk)
            page_metadata.append({
                "page": page_number
            })

print("Total chunks:", len(page_chunks))
print("First chunk page:", page_metadata[0]["page"])
print("First chunk preview:", page_chunks[0][:200])
```

```python
# Day 2 — Step 2: Store chunks with page metadata

# 1. Generate embeddings for our new chunks
page_embeddings = model.encode(
    page_chunks,
    show_progress_bar=True
)

# 2. Create a new ChromaDB collection
page_collection = client.get_or_create_collection(
    name="documind_pages"
)

# 3. Store chunks, embeddings, and page numbers
page_collection.upsert(
    ids=[f"page_chunk_{i}" for i in range(len(page_chunks))],
    documents=page_chunks,
    embeddings=page_embeddings.tolist(),
    metadatas=page_metadata
)

# 4. Verify storage
print("Stored chunks:", page_collection.count())
```

```python
# Day 2 — Step 3: Retrieval with page numbers

question = "How can family trauma affect future generations?"

# Convert the question into an embedding
question_embedding = model.encode(question)

# Search the new ChromaDB collection
results = page_collection.query(
    query_embeddings=[question_embedding.tolist()],
    n_results=3,
    include=["documents", "metadatas", "distances"]
)

# Display the retrieved passages and their pages
for i in range(len(results["documents"][0])):
    text = results["documents"][0][i]
    page = results["metadatas"][0][i]["page"]
    distance = results["distances"][0][i]

    print(f"\n--- Result {i + 1} | PDF Page {page} ---")
    print(f"Distance: {distance:.4f}")
    print(text[:300])
```

```python
# Day 2 — Step 4: RAG with source citations

# 1. Build context with page references
context_parts = []

for text, metadata in zip(
    results["documents"][0],
    results["metadatas"][0]
):
    page = metadata["page"]
    context_parts.append(f"[PDF Page {page}]\n{text}")

context = "\n\n".join(context_parts)

# 2. Create the prompt
prompt = f"""
You are DocuMind AI, a document assistant.

Instructions:
- Answer using ONLY the provided PDF context.
- Cite the PDF page number for each important claim.
- If the context does not contain the answer,
  say "I don't know based on the document."
- Do not invent information or page numbers.

Context:
{context}

Question:
{question}

Answer:
"""

# 3. Generate the answer
response = ollama.chat(
    model="llama3.2:3b",
    messages=[
        {"role": "user", "content": prompt}
    ]
)

# 4. Display the answer
print(response["message"]["content"])
```

```python
# Day 3 — Step 2: Relevance filtering

def retrieve_relevant(question, threshold=1.0, top_k=3):

    question_embedding = model.encode(question)

    results = page_collection.query(
        query_embeddings=[question_embedding.tolist()],
        n_results=top_k,
        include=["documents", "metadatas", "distances"]
    )

    relevant_chunks = []

    for text, metadata, distance in zip(
        results["documents"][0],
        results["metadatas"][0],
        results["distances"][0]
    ):
        if distance <= threshold:
            relevant_chunks.append({
                "text": text,
                "page": metadata["page"],
                "distance": distance
            })

    return relevant_chunks
```

```python
# Day 3 — Step 4: Filtered RAG pipeline

def ask_documind(question):

    # 1. Retrieve relevant chunks
    chunks = retrieve_relevant(
        question,
        threshold=1.0,
        top_k=5
    )

    # 2. Stop if no chunks pass the filter
    if not chunks:
        return "I don't know based on the document."

    # 3. Build context with PDF page numbers
    context = "\n\n".join(
        f"[PDF Page {chunk['page']}]\n{chunk['text']}"
        for chunk in chunks
    )

    # 4. Create the prompt
    prompt = f"""
You are DocuMind AI.

Answer the question using ONLY the provided context.
Cite the PDF page numbers supporting your answer.
If the context does not support an answer, say:
"I don't know based on the document."

Context:
{context}

Question:
{question}
"""

    # 5. Generate the answer
    response = ollama.chat(
        model="llama3.2:3b",
        messages=[
            {"role": "user", "content": prompt}
        ]
    )

    return response["message"]["content"]
```

```python
# Day 4 — Step 5: Semantic + keyword retrieval

def hybrid_retrieve(question, keyword, top_k=10):

    question_embedding = model.encode(question)

    results = page_collection.query(
        query_embeddings=[question_embedding.tolist()],
        n_results=top_k,
        include=["documents", "metadatas", "distances"]
    )

    ranked_chunks = []

    for text, metadata, distance in zip(
        results["documents"][0],
        results["metadatas"][0],
        results["distances"][0]
    ):
        keyword_match = keyword.lower() in text.lower()

        ranked_chunks.append({
            "page": metadata["page"],
            "distance": distance,
            "keyword_match": keyword_match,
            "text": text
        })

    # Prioritize keyword matches, then semantic distance
    ranked_chunks.sort(
        key=lambda chunk: (
            not chunk["keyword_match"],
            chunk["distance"]
        )
    )

    return ranked_chunks[:5]


chunks = hybrid_retrieve(
    question="What role does Bert Hellinger play in the book?",
    keyword="Hellinger"
)

for chunk in chunks:
    print(
        f"Page {chunk['page']} | "
        f"Distance: {chunk['distance']:.4f} | "
        f"Keyword match: {chunk['keyword_match']}"
    )
```

```python
# Restore DocuMind AI after restarting Jupyter

import pymupdf
import chromadb
import ollama
from sentence_transformers import SentenceTransformer

# Load our PDF
document = pymupdf.open("sample.pdf")

# Load our embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Connect to our saved ChromaDB database
client = chromadb.PersistentClient(path="./chroma_db")

# Open our saved collection
page_collection = client.get_collection(
    name="documind_pages"
)

print("PDF pages:", len(document))
print("Stored chunks:", page_collection.count())
```

```python
# Day 5 — Step 4: Restore hybrid retrieval

def hybrid_retrieve(question, keyword, top_k=10):

    question_embedding = model.encode(question)

    results = page_collection.query(
        query_embeddings=[question_embedding.tolist()],
        n_results=top_k,
        include=["documents", "metadatas", "distances"]
    )

    ranked_chunks = []

    for text, metadata, distance in zip(
        results["documents"][0],
        results["metadatas"][0],
        results["distances"][0]
    ):
        keyword_match = keyword.lower() in text.lower()

        ranked_chunks.append({
            "page": metadata["page"],
            "distance": distance,
            "keyword_match": keyword_match,
            "text": text
        })

    ranked_chunks.sort(
        key=lambda chunk: (
            not chunk["keyword_match"],
            chunk["distance"]
        )
    )

    return ranked_chunks[:5]


print("Hybrid retrieval restored!")

```

```python
# Day 5 — Step 5: Restore RAG pipeline

def ask_documind_v2(question, keyword):

    # 1. Retrieve relevant chunks
    chunks = hybrid_retrieve(
        question=question,
        keyword=keyword,
        top_k=10
    )

    # 2. Filter out distant chunks
    chunks = [
        chunk for chunk in chunks
        if chunk["distance"] <= 1.2
    ]

    # 3. Handle questions without relevant context
    if not chunks:
        return "I don't know based on the document."

    # 4. Build context with page references
    context = "\n\n".join(
        f"[PDF Page {chunk['page']}]\n{chunk['text']}"
        for chunk in chunks
    )

    # 5. Prepare the prompt
    prompt = f"""
You are DocuMind AI.

Answer using ONLY the provided document context.
Cite the PDF page numbers supporting your answer.
Distinguish the author's claims from established facts.
If the answer is unsupported, say:
"I don't know based on the document."

Context:
{context}

Question: {question}
"""

    # 6. Generate the answer locally
    response = ollama.chat(
        model="llama3.2:1b",
        messages=[
            {"role": "user", "content": prompt}
        ],
        options={"num_ctx": 2048}
    )

    return response["message"]["content"]


print("DocuMind AI restored!")
```

```python
# Day 5 — Step 8: Citation validation

import re

def validate_citations(answer, chunks):

    allowed_pages = {
        chunk["page"] for chunk in chunks
    }

    cited_pages = {
        int(page)
        for page in re.findall(
            r"(?:PDF\s+)?Page\s+(\d+)",
            answer,
            flags=re.IGNORECASE
        )
    }

    invalid_pages = cited_pages - allowed_pages

    print("Available pages:", sorted(allowed_pages))
    print("Cited pages:", sorted(cited_pages))
    print("Invalid citations:", sorted(invalid_pages))

    return len(invalid_pages) == 0
```

```python
is_valid = validate_citations(answer, chunks)

print("Citation page check passed:", is_valid)
```
